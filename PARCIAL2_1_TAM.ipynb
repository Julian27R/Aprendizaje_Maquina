#**4. Crear archivo dashboard_USPS.py para visualización con Streamlit**

```python
%%writefile app.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
from sklearn.metrics import ConfusionMatrixDisplay, roc_auc_score, RocCurveDisplay, roc_curve, classification_report, accuracy_score, f1_score
from sklearn.preprocessing import label_binarize


# --- CARGA DE DATOS ---
y_test       = np.load("y_test.npy")
X_pca        = np.load("X_pca.npy")
X_umap_k5    = np.load("X_umap_k5.npy")
X_umap_k15   = np.load("X_umap_k15.npy")
X_umap_k50   = np.load("X_umap_k50.npy")
y_pred_knn   = np.load("y_pred_knn.npy")
proba_knn    = np.load("proba_knn.npy")
y_pred_svc   = np.load("y_pred_svc.npy")
proba_svc    = np.load("proba_svc.npy")
y_pred_cnn   = np.load("y_pred_cnn.npy")
proba_cnn    = np.load("proba_cnn.npy")
y_train      = np.load("y_train.npy")
X_train      = np.load("X_train.npy")

# Subconjunto usado para PCA y UMAP
y_train_small = y_train[:3000]
X_train_small = X_train[:3000]

# --- FORMULAS DE MODELOS ---
def show_model_formulas():
    st.markdown("## 🧠 Modelos y Problemas de Optimización")

    with st.expander("🔹 1. PCA – Análisis de Componentes Principales"):
        st.markdown(r"""
**Descripción:** técnica de reducción de dimensionalidad que proyecta datos de alta dimensión a un subespacio de menor dimensión **maximizando la varianza**.

**Modelo generativo:**

$$
X = ZW^{\top}, \quad Z = XW
$$

**Optimización (mínima reconstrucción):**

$$
\min_W \;\|X - XWW^{\top}\|_F^2 \quad \text{s.t. } WW^{\top} = I
$$

**Perspectiva geométrica (máxima varianza):**

$$
\max_W \;\operatorname{tr}(W^{\top} \Sigma W) \quad \text{s.t. } WW^{\top} = I
$$

Donde $\Sigma = \frac{1}{N} X^\top X$.
""")

    with st.expander("🔹 2. UMAP – Uniform Manifold Approximation and Projection"):
        st.markdown(r"""
**Probabilidad en espacio original:**

$$
p_{nm} = \exp\left(-\frac{\|x_n - x_m\| - \rho_n}{\sigma_n}\right)
$$

**Simetrización del grafo:**

$$
p'_{nm} = p_{nm} + p_{mn} - p_{nm}p_{mn}
$$

**Probabilidad en espacio reducido:**

$$
q_{nm} = \left(1 + a\|z_n - z_m\|^{2b}\right)^{-1}
$$

**Función objetivo: entropía cruzada**

$$
C(P, Q) =
\sum_{n,m} \left[ -p'_{nm} \log q_{nm} - (1 - p'_{nm}) \log (1 - q_{nm}) \right]
$$
""")

    with st.expander("🔹 3. KNN – K-Nearest Neighbors"):
        st.markdown(r"""
**Predicción por mayoría:**

$$
\hat{y}(x) = \operatorname{mode} \left\{ y_i : x_i \in N_K(x) \right\}
$$

**Forma alternativa:**

$$
\hat{y}(x) = \arg\max_c \sum_{x_i \in N_K(x)} \mathbf{1}(y_i = c)
$$
""")

    with st.expander("🔹 4. SVC – Support Vector Classifier"):
        st.markdown(r"""
**Modelo lineal:**

$$
f(x) = W^\top x + b, \quad \hat{y} = \operatorname{sign}(f(x))
$$

**Problema (separable):**

$$
\min_{W,b} \tfrac{1}{2} \|W\|^2 \quad \text{s.t. } y_i(W^\top x_i + b) \geq 1
$$

**Problema (margen suave):**

$$
\min_{W,b,\{\xi_i\}} \tfrac{1}{2} \|W\|^2 + C\sum_i \xi_i \quad
\text{s.t. } y_i(W^\top x_i + b) \geq 1 - \xi_i,\quad \xi_i \ge 0
$$
""")

    with st.expander("🔹 5. CNN – Convolutional Neural Network"):
        st.markdown(r"""
**Clasificador convolucional:**

$$
f_\theta : \mathbb{R}^{1\times16\times16} \to \{0,1,\dots,9\}
$$

**Salida softmax:**

$$
\hat{y}_i = \operatorname{softmax}(f_\theta(x_i)) \in [0,1]^{10}
$$

**Función de pérdida: entropía cruzada**

$$
\theta^* = \arg\min_\theta \left[
- \sum_{i=1}^N \log \hat{y}_{i,y_i}
\right]
=
\arg\min_\theta \left[
- \sum_{i=1}^N \log\left(
\frac{e^{z_{i,y_i}}}{\sum_{c=1}^{10} e^{z_{i,c}}}
\right)
\right]
$$

**Optimización:** AdamW + scheduler ReduceLROnPlateau

**Evaluación:**
- Accuracy
- F1 macro
- Matriz de confusión
- Curva ROC macro-promediada
""")


# --- PLOT EMBEDDING ---
def plot_embedding(ax, emb, y_vals, title, X_imgs):
    ax.scatter(emb[:, 0], emb[:, 1], c=y_vals, cmap="tab10", s=4, alpha=0.6, linewidths=0)
    ax.set_title(title)
    ax.set_xticks([]); ax.set_yticks([])
    for digit in np.unique(y_vals):
        idx = np.where(y_vals == digit)[0]
        if idx.size == 0: continue
        med = idx[len(idx)//2]
        xy = emb[med]
        img = X_imgs[med].reshape(16, 16)
        thumb = OffsetImage(img, cmap="gray", zoom=1.5)
        ax.add_artist(AnnotationBbox(thumb, xy, frameon=False))

# --- PLOT CONFUSION Y ROC ---
def plot_confusion_roc(y_true, y_pred, y_score, model_name):
    fig, axs = plt.subplots(1, 2, figsize=(10, 4))
    disp = ConfusionMatrixDisplay.from_predictions(
        y_true, y_pred, normalize='true', values_format='.2f',
        cmap='Blues', ax=axs[0], colorbar=False)
    axs[0].set_title(f"{model_name} – Confusion Matrix")
    for text in disp.text_.ravel(): text.set_fontsize(7)
    y_bin = label_binarize(y_true, classes=np.arange(10))
    auc = roc_auc_score(y_bin, y_score, average='macro')
    RocCurveDisplay.from_predictions(
        y_bin.ravel(), y_score.ravel(), ax=axs[1],
        name=f"AUC {auc:.3f}", linewidth=1.5)
    axs[1].plot([0,1], [0,1], 'k--', lw=1)
    axs[1].set_title(f"{model_name} – ROC macro")
    st.pyplot(fig)

# --- PLOT ROC POR CLASE ---
def plot_rocs_per_class(y_true, y_score, model_name):
    n_classes = y_score.shape[1]
    y_bin = label_binarize(y_true, classes=np.arange(n_classes))

    plt.figure(figsize=(8, 6))
    for i in range(n_classes):
        fpr, tpr, _ = roc_curve(y_bin[:, i], y_score[:, i])
        auc_i = roc_auc_score(y_bin[:, i], y_score[:, i])
        plt.plot(fpr, tpr, lw=2, label=f"Clase {i} (AUC = {auc_i:.3f})")

    # Diagonal aleatoria
    plt.plot([0, 1], [0, 1], 'k--', lw=1)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel("Tasa de Falsos Positivos (FPR)")
    plt.ylabel("Tasa de Verdaderos Positivos (TPR)")
    plt.title(f"Curvas ROC por clase — {model_name}")
    plt.legend(loc="lower right", fontsize=8)
    plt.grid(True)
    st.pyplot(plt.gcf())


# --- MAIN DASHBOARD ---
def main():
    st.title("🔍 Dashboard Interactivo — USPS")
    st.sidebar.title("Opciones del dashboard")
    section = st.sidebar.radio("Selecciona una sección:", ["Formulación", "Visualización PCA/UMAP", "Resultados de modelos"])

    if section == "Formulación":
        show_model_formulas()

    elif section == "Visualización PCA/UMAP":
        st.subheader("🔍 Proyecciones en 2D")
        selected_digits = st.multiselect("Selecciona las clases a visualizar", list(range(10)), default=list(range(10)))
        tipo_proj = st.selectbox("Selecciona tipo de proyección:", ["PCA", "UMAP (k=5)", "UMAP (k=15)", "UMAP (k=50)", "Todas"])

        if not selected_digits:
            st.warning("Selecciona al menos una clase.")
        else:
            mask = np.isin(y_train_small, selected_digits)
            fig, axs = None, None

            if tipo_proj == "Todas":
                fig, axs = plt.subplots(2, 2, figsize=(12, 10))
                plot_embedding(axs[0,0], X_pca[mask], y_train_small[mask], "PCA (2-D)", X_train_small[mask])
                plot_embedding(axs[0,1], X_umap_k5[mask], y_train_small[mask], "UMAP (k=5)", X_train_small[mask])
                plot_embedding(axs[1,0], X_umap_k15[mask], y_train_small[mask], "UMAP (k=15)", X_train_small[mask])
                plot_embedding(axs[1,1], X_umap_k50[mask], y_train_small[mask], "UMAP (k=50)", X_train_small[mask])
            else:
                fig, ax = plt.subplots(figsize=(6, 5))
                if tipo_proj == "PCA":
                    plot_embedding(ax, X_pca[mask], y_train_small[mask], "PCA (2-D)", X_train_small[mask])
                elif tipo_proj == "UMAP (k=5)":
                    plot_embedding(ax, X_umap_k5[mask], y_train_small[mask], "UMAP (k=5)", X_train_small[mask])
                elif tipo_proj == "UMAP (k=15)":
                    plot_embedding(ax, X_umap_k15[mask], y_train_small[mask], "UMAP (k=15)", X_train_small[mask])
                elif tipo_proj == "UMAP (k=50)":
                    plot_embedding(ax, X_umap_k50[mask], y_train_small[mask], "UMAP (k=50)", X_train_small[mask])
            plt.tight_layout()
            st.pyplot(fig)

    elif section == "Resultados de modelos":
        model = st.selectbox("Elige modelo", ["KNN", "SVC", "CNN"])
        show_all_rocs = st.checkbox("Mostrar ROC por clase (OvR)")

        if model == "KNN":
            st.markdown("### 🔹 KNN – Métricas")
            st.text(classification_report(y_test, y_pred_knn, digits=4))
            plot_confusion_roc(y_test, y_pred_knn, proba_knn, "KNN")
            if show_all_rocs:
                plot_rocs_per_class(y_test, proba_knn, "KNN")

        elif model == "SVC":
            st.markdown("### 🔹 SVC – Métricas")
            st.text(classification_report(y_test, y_pred_svc, digits=4))
            plot_confusion_roc(y_test, y_pred_svc, proba_svc, "SVC")
            if show_all_rocs:
                plot_rocs_per_class(y_test, proba_svc, "SVC")

        elif model == "CNN":
            st.markdown("### 🔹 CNN – Métricas")
            st.text(classification_report(y_test, y_pred_cnn, digits=4))
            plot_confusion_roc(y_test, y_pred_cnn, proba_cnn, "CNN")
            if show_all_rocs:
                plot_rocs_per_class(y_test, proba_cnn, "CNN")
if __name__ == "__main__":
    main()
```
